```python
control_run1 = np.load("control_history_run1.npy")
control_run2 = np.load("control_history_run2.npy")

print("Run 1 (Control):", control_run1)
print("Run 2 (Control):", control_run2)

if control_run1.shape != control_run2.shape:
    print(f"Cannot compare Control runs: Shape mismatch {control_run1.shape} vs {control_run2.shape}")
    print("Reproducible: False (Empty or incomplete data)")
else:
    print(
        "Reproducible:",
        np.allclose(control_run1, control_run2, atol=1e-6, rtol=0)
    )

print("-" * 40)

shallow_run1 = np.load("shallow_history_run1.npy")
shallow_run2 = np.load("shallow_history_run2.npy")

print("Run 1 (Shallow):", shallow_run1)
print("Run 2 (Shallow):", shallow_run2)

if shallow_run1.shape != shallow_run2.shape:
    print(f"Cannot compare Shallow runs: Shape mismatch {shallow_run1.shape} vs {shallow_run2.shape}")
    print("Reproducible: False (Empty or incomplete data)")
else:
    print(
        "Reproducible:",
        np.allclose(shallow_run1, shallow_run2, atol=1e-6, rtol=0)
    )
```
